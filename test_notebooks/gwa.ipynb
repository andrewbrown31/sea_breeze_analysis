```python
import matplotlib.pyplot as plt
import xarray as xr
import rioxarray
import numpy as np
```

```python
#Load the .tif files using rioxarray

gwa_cf_class2 = rioxarray.open_rasterio("/g/data/ng72/ab4502/gwa/AUS_capacity-factor_IEC2.tif")
gwa_cf_class3 = rioxarray.open_rasterio("/g/data/ng72/ab4502/gwa/AUS_capacity-factor_IEC3.tif")
gwa_ws = rioxarray.open_rasterio("/g/data/ng72/ab4502/gwa/AUS_wind-speed_100m.tif")
```

```python
#Interpolate to a coraser grid - otherwise it takes a long time to plot
interp_lat = np.arange(-50,-7.75,0.25)
interp_lon = np.arange(109,163.25,0.25)

gwa_cf_class2_interp = gwa_cf_class2.interp({"x":interp_lon,"y":interp_lat})
gwa_cf_class3_interp = gwa_cf_class3.interp({"x":interp_lon,"y":interp_lat})
gwa_ws_interp = gwa_ws.interp({"x":interp_lon,"y":interp_lat})
```

```python
plt.figure(figsize=[12,2])
plt.subplot(1,3,1)
gwa_cf_class2_interp.squeeze().plot()

plt.subplot(1,3,2)
gwa_cf_class3_interp.squeeze().plot()

plt.subplot(1,3,3)
gwa_ws_interp.squeeze().plot()
```
